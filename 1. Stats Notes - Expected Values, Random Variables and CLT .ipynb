# Randomness and Law of Large Numbers

```python
# A biased coin: setting the probability of heads to 70%
# Heads is encoded as 1, tails as 0.
p_heads = 0.7


# 4 different numbers of trials (coin flips) to compare how results behave as sample size increases.
flip_sizes = [3, 10, 50, 100, 1000, 100000]



print("Biased coin: probability of heads =", p_heads)
print()


for n in flip_sizes:
  # Draw n independent samples, from the set {1, 0} with probabilities {0.7, 0.3} respectively
  flips = np.random.choice([1, 0], size=n, p=[p_heads, 1 - p_heads])

  avg = flips.mean()
  print(f"Flips: {n:>6} | Average outcome: {avg:.4f}")
```

## What are we trying to simulate?

We simulate coin flips with a fixed true probability (0.7) and use the sample average as an estimate of that probability. By increasing sample size, we observe how estimation uncertainty decreases as more data is collected.

This demonstrates that probability is a property of the **data-generating process**, and the sample mean is an estimator of that underlying rate.


## Assumptions

The simulation assumes:

* **Independence:** each flip does not affect the next
* **Identical distribution:** the success rate stays constant over time

Together, this is **IID data (Independent and Identically Distributed)**.

Most statistical models quietly assume IID. In real systems this is rarely perfectly true (seasonality, feedback loops, user behavior changes), which is why models often degrade in production.


## What are we observing?

### Small samples → unstable estimates

* Each data point has large influence
* Randomness dominates signal
* High variance

### Large samples → stable estimates

* Each new data point has tiny influence
* Randomness averages out
* Variance decreases

The coin itself does not become less random. What changes is the precision of our estimate: as more observations are averaged together, randomness cancels out and the sample mean converges toward the true probability. This is an example of the **Law of Large Numbers**.


## Terms to Note

* **Estimator:** the sample mean is an estimator of the true probability
* **Variance reduction:** averaging many noisy samples produces a more stable result
* **Signal vs noise:** structure emerges only at scale
* **Law of Large Numbers:** sample averages converge to expected values
* **Sampling uncertainty:** small datasets are inherently unreliable
* **Data-generating process:** models learn processes, not individual outcomes


## Key Takeaways:
* Applying this principle to ML: Models don’t predict single events well. They estimate long-run behavior across many similar cases.
* Randomness exists at the individual level, but structure emerges at scale. Statistics works because averages reveal the underlying process even when individual outcomes are unpredictable.
